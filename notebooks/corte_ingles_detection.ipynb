```python
# 1. Instalar dependencias
!pip install ultralytics roboflow

# Verificar versiones de las librerías
import ultralytics
import roboflow

print(f"Ultralytics version: {ultralytics.__version__}")
print(f"Roboflow version: {roboflow.__version__}")

# 2. Importar el dataset
from roboflow import Roboflow

try:
    rf = Roboflow(api_key="dfoJRMbUQTtgVJ1ls7CJ")
    project = rf.workspace("corteingleslogodetection").project("corteingles_logo_detection")
    version = project.version(4)
    dataset = version.download("yolov8-obb")
    print("Dataset descargado exitosamente.")
except Exception as e:
    print(f"Error al descargar el dataset: {e}")

# 3. Verificar GPU
import torch

if torch.cuda.is_available():
    print(f"GPU disponible: {torch.cuda.get_device_name(0)}")
    device = 0  # Usar GPU
else:
    print("GPU no disponible, se utilizará CPU.")
    device = 'cpu'

# 4. Importar YOLO y configurar entrenamiento
from ultralytics import YOLO

# Inicializar modelo YOLOv8
model = YOLO('yolov8n.yaml')  # YOLOv8n es ligero y adecuado para datasets pequeños

# Entrenar el modelo
results = model.train(
    data=f'{dataset.location}/data.yaml',  # Ruta al archivo data.yaml del dataset
    epochs=100,
    imgsz=300,  # Tamaño nativo de las imágenes
    batch=16,  # Tamaño moderado compatible con GPU
    name='corte_ingles_v4',  # Nombre del experimento
    device=device,  # GPU o CPU
    patience=10,  # Detener temprano si no mejora en 10 epochs
    lr0=0.01,  # Tasa de aprendizaje inicial
    lrf=0.2,  # Mayor reducción final de lr
    momentum=0.937,  # Momentum
    weight_decay=0.0005,  # Regularización L2
    warmup_epochs=3,  # Epochs de calentamiento
    warmup_momentum=0.8,  # Momentum durante el calentamiento
    warmup_bias_lr=0.1  # Tasa de aprendizaje del bias durante calentamiento
)

# Visualizar resultados del entrenamiento
results.plot()

# 5. Guardar y descargar los pesos entrenados
import os
from google.colab import files

output_dir = '/content/runs/detect/corte_ingles_v4/weights/'
best_weights_path = os.path.join(output_dir, 'best.pt')
last_weights_path = os.path.join(output_dir, 'last.pt')

# Verificar y descargar el archivo best.pt
if os.path.exists(best_weights_path):
    files.download(best_weights_path)
    print("best.pt descargado correctamente.")
else:
    print("Error: best.pt no encontrado.")

# Verificar y descargar el archivo last.pt
if os.path.exists(last_weights_path):
    files.download(last_weights_path)
    print("last.pt descargado correctamente.")
else:
    print("Error: last.pt no encontrado.")

# 6. Liberar memoria
torch.cuda.empty_cache()

# 7. Visualizar métricas en TensorBoard
%load_ext tensorboard
%tensorboard --logdir /content/runs/detect/corte_ingles_v4
```
